```python
# Load task-affordance data and bundle scores for the occupational analysis dataset.
from pathlib import Path
import pandas as pd

ROOT = Path("/content/drive/MyDrive/genai-human-skills-at-work/Occupational-level")

affordance = pd.read_csv(
    ROOT / "01_Source_Data_Preparation" / "Input" / "Occupational_Task_Affordance_Configuration_v6.csv"
)
bundle_scores = pd.read_csv(
    ROOT / "03_Skill_Bundle_Construction" / "Output" / "Occupational_Skill_Bundle_Scores.csv"
)
```

```python
# Merge bundle scores into the RQ1 dataset and display matching coverage.
analysis_data = affordance.merge(
    bundle_scores,
    on="soc6",
    how="left",
    validate="one_to_one",
    indicator=True
)
analysis_data["_merge"].value_counts()
```

```python
# Retain occupations with both task-affordance data and skill-bundle scores.
analysis_data = analysis_data.loc[
    analysis_data["_merge"] == "both"
].drop(columns="_merge")

print(f"Matched occupations retained: {len(analysis_data)}")
```

```python
# Save the matched occupational dataset for Phase 05 analysis.
analysis_data.to_csv(
    ROOT / "04_Analysis_Dataset" / "Output" / "Occupational_Analysis_Dataset_Full.csv",
    index=False
)
```

### **RQ3a: Access and meeting the deadline Access Yes or No, with an assessable met/missed deadline**

```python
# Create the RQ3a working dataset
rq3a = df_filtered.iloc[:, [0, 1, 2, 3, 4, 5, 6, 7, 9]].copy()

rq3a.columns = [
    "sector",
    "role",
    "role_tenure",
    "restriction",
    "approved_access",
    "task_type",
    "difficulty",
    "task_status",
    "deadline"
]

# Inspect the original responses before coding
for column in ["restriction", "approved_access", "deadline"]:
    print(f"\n{column}")
    print(rq3a[column].value_counts(dropna=False).to_string())
```

```python
import pandas as pd

rq3a["access_yes"] = (
    rq3a["approved_access"].astype("string").str.strip()
    .map({"Yes": 1, "No": 0})
    .astype("Int64")
)

rq3a["deadline_met"] = (
    rq3a["deadline"].astype("string").str.strip()
    .map({
        "The task was completed on or before the deadline": 1,
        "The deadline passed, and the task was unfinished at that point": 0
    })
    .astype("Int64")
)

rq3a_analysis = rq3a.dropna(
    subset=["access_yes", "deadline_met"]
).copy()

print(f"RQ3a comparison sample: {len(rq3a_analysis)}")

display(
    pd.crosstab(
        rq3a_analysis["access_yes"],
        rq3a_analysis["deadline_met"]
    )
    .reindex(index=[0, 1], columns=[0, 1], fill_value=0)
    .rename(index={0: "No access", 1: "Access"},
            columns={0: "Missed", 1: "Met"})
)
```

```python
from scipy.stats import binomtest, fisher_exact
from math import sqrt

# Counts from our RQ3a dataset
rq3a_counts = (
    rq3a_analysis.groupby("access_yes")["deadline_met"]
    .agg(["sum", "count"])
)
m1, n1 = map(int, rq3a_counts.loc[1])  # Access
m0, n0 = map(int, rq3a_counts.loc[0])  # No access

# Difference in on-time rates
p1, p0 = m1 / n1, m0 / n0
difference = p1 - p0

# Newcombe 95% confidence interval using Wilson limits
l1, u1 = binomtest(m1, n1).proportion_ci(method="wilson")
l0, u0 = binomtest(m0, n0).proportion_ci(method="wilson")

lower = difference - sqrt((p1 - l1)**2 + (u0 - p0)**2)
upper = difference + sqrt((u1 - p1)**2 + (p0 - l0)**2)

# Two-sided Fisher's exact test
_, p_value = fisher_exact(
    [[m1, n1 - m1], [m0, n0 - m0]],
    alternative="two-sided"
)

print(f"Difference: {100 * difference:.1f} percentage points")
print(f"95% CI: {100 * lower:.1f} to {100 * upper:.1f} percentage points")
print(f"Fisher's exact p-value: {p_value:.4f}")
```

```python
for column in ["difficulty", "task_type"]:
    categories = (
        rq3a_analysis[column].astype("string")
        .str.strip().replace("", pd.NA).fillna("Missing")
    )

    counts = pd.crosstab(
        categories, rq3a_analysis["access_yes"]
    ).reindex(columns=[0, 1], fill_value=0)

    summary = counts.astype(str)

    for group in [0, 1]:
        percentages = 100 * counts[group] / counts[group].sum()
        summary[group] = (
            counts[group].astype(str)
            + " (" + percentages.round(1).astype(str) + "%)"
        )

    print(f"\n{column}")
    display(summary.rename(columns={0: "No access", 1: "Access"}))
```

```python
# Deadline outcomes within each difficulty and access group
difficulty = (
    rq3a_analysis["difficulty"].astype("string").str.strip()
)

rq3a_difficulty_outcomes = pd.crosstab(
    index=[difficulty, rq3a_analysis["access_yes"]],
    columns=rq3a_analysis["deadline_met"]
).reindex(columns=[0, 1], fill_value=0)

rq3a_difficulty_outcomes = (
    rq3a_difficulty_outcomes
    .rename(columns={0: "Missed", 1: "Met"})
    .rename(index={0: "No access", 1: "Access"}, level="access_yes")
)

display(rq3a_difficulty_outcomes)
```

```python
task_type = (
    rq3a_analysis["task_type"].astype("string").str.strip()
)

rq3a_task_outcomes = pd.crosstab(
    index=[task_type, rq3a_analysis["access_yes"]],
    columns=rq3a_analysis["deadline_met"]
).reindex(columns=[0, 1], fill_value=0)

rq3a_task_outcomes = (
    rq3a_task_outcomes
    .rename(columns={0: "Missed", 1: "Met"})
    .rename(index={0: "No access", 1: "Access"}, level="access_yes")
)

display(rq3a_task_outcomes)
```

### **RQ3b: Access and hands-on time	Access Yes or No, task completed, and usable minutes**

```python
# Prepare RQ3b from the full retained dataset
rq3b = df_filtered.iloc[:, [4, 5, 6, 7, 8]].copy()

rq3b.columns = [
    "approved_access",
    "task_type",
    "difficulty",
    "task_status",
    "minutes_raw"
]

for column in ["approved_access", "task_status"]:
    rq3b[column] = rq3b[column].astype("string").str.strip()

print("Task status — all retained responses:")
print(rq3b["task_status"].value_counts(dropna=False).to_string())

# Select completed tasks independently of deadline outcome
rq3b_completed = rq3b.loc[
    rq3b["task_status"].eq("Completed")
].copy()

print(f"\nCompleted tasks: {len(rq3b_completed)}")

for column in ["approved_access", "minutes_raw"]:
    print(f"\n{column}")
    print(rq3b_completed[column].value_counts(dropna=False).to_string())
```

```python
rq3b_analysis = rq3b_completed.copy()

# Standardize units while preserving the original responses
time_text = (
    rq3b_analysis["minutes_raw"].astype("string")
    .str.strip().str.lower()
    .str.replace(r"\s*minutes?$", "", regex=True)
)

rq3b_analysis["minutes_lower"] = pd.to_numeric(
    time_text.replace({"90-120": "90"}), errors="raise"
)
rq3b_analysis["minutes_upper"] = pd.to_numeric(
    time_text.replace({"90-120": "120"}), errors="raise"
)

# Working estimate; exact entries have identical lower/upper bounds
rq3b_analysis["minutes"] = (
    rq3b_analysis["minutes_lower"]
    + rq3b_analysis["minutes_upper"]
) / 2

print("Range response:")
display(rq3b_analysis.loc[
    time_text.eq("90-120"),
    ["approved_access", "minutes_raw", "minutes"]
])

print("Working-time summaries, in minutes:")
display(
    rq3b_analysis.groupby("approved_access")["minutes"]
    .agg(
        n="size",
        median="median",
        q1=lambda x: x.quantile(0.25),
        q3=lambda x: x.quantile(0.75),
        mean="mean",
        minimum="min",
        maximum="max"
    )
    .reindex(["No", "Yes"])
    .round(1)
)
```

```python
import numpy as np
from scipy.stats import bootstrap

def mean_difference(access, no_access, axis=-1):
    return np.mean(access, axis=axis) - np.mean(no_access, axis=axis)

results = []

for label, column in [
    ("105 minutes — midpoint", "minutes"),
    ("90 minutes — lower bound", "minutes_lower"),
    ("120 minutes — upper bound", "minutes_upper"),
]:
    access = rq3b_analysis.loc[
        rq3b_analysis["approved_access"].eq("Yes"), column
    ].to_numpy(dtype=float)

    no_access = rq3b_analysis.loc[
        rq3b_analysis["approved_access"].eq("No"), column
    ].to_numpy(dtype=float)

    ci = bootstrap(
        (access, no_access),
        mean_difference,
        paired=False,
        vectorized=True,
        method="BCa",
        confidence_level=0.95,
        n_resamples=20000,
        random_state=2026
    ).confidence_interval

    results.append({
        "Range response coded as": label,
        "Mean difference (Yes − No)": mean_difference(access, no_access),
        "95% CI lower": ci.low,
        "95% CI upper": ci.high
    })

rq3b_time_comparison = pd.DataFrame(results)
display(rq3b_time_comparison.round(1))
```

```python
for column in ["task_type", "difficulty"]:
    categories = (
        rq3b_analysis[column]
        .astype("string")
        .str.strip()
        .fillna("Missing")
    )

    counts = pd.crosstab(
        categories,
        rq3b_analysis["approved_access"]
    ).reindex(columns=["No", "Yes"], fill_value=0)

    percentages = counts.div(counts.sum(axis=0), axis=1) * 100

    table = (
        counts.astype(str)
        + " ("
        + percentages.round(1).astype(str)
        + "%)"
    )

    print(f"\n{column} — completed tasks:")
    display(table.rename(columns={
        "No": "No approved access",
        "Yes": "Approved access"
    }))
```

```python
rq3b_time_by_category = {}

for column in ["difficulty", "task_type"]:
    summary = (
        rq3b_analysis
        .groupby([column, "approved_access"], dropna=False)["minutes"]
        .agg(
            n="size",
            mean="mean",
            median="median",
            maximum="max"
        )
        .round(1)
    )

    rq3b_time_by_category[column] = summary

    print(f"\nTime in minutes by {column} and approved access:")
    display(summary)
```

```python
import statsmodels.formula.api as smf

rq3b_model_data = rq3b_analysis.copy()

rq3b_model_data["access_yes"] = (
    rq3b_model_data["approved_access"]
    .map({"No": 0, "Yes": 1})
    .astype(float)
)

rq3b_model_data["difficulty"] = (
    rq3b_model_data["difficulty"].astype(str).str.strip()
)
rq3b_model_data["minutes"] = (
    rq3b_model_data["minutes"].astype(float)
)

rq3b_difficulty_model = smf.ols(
    "minutes ~ access_yes + C(difficulty)",
    data=rq3b_model_data,
    missing="raise"
).fit(cov_type="HC3", use_t=True)

estimate = rq3b_difficulty_model.params["access_yes"]
lower, upper = rq3b_difficulty_model.conf_int().loc["access_yes"]

print(f"Completed tasks included: {int(rq3b_difficulty_model.nobs)}")
print(f"Difficulty-adjusted difference (Yes − No): {estimate:.1f} minutes")
print(f"Approximate 95% CI: {lower:.1f} to {upper:.1f} minutes")
```

```python
import numpy as np
import statsmodels.formula.api as smf

data = rq3b_model_data.copy()
data["task_type"] = data["task_type"].astype(str).str.strip()

# Retain task categories represented in both access groups
counts = pd.crosstab(data["task_type"], data["access_yes"])
shared_types = counts.index[counts.gt(0).all(axis=1)]
rq3b_shared = data.loc[data["task_type"].isin(shared_types)].copy()

specifications = {
    "Unadjusted": "minutes ~ access_yes",
    "Difficulty adjusted": "minutes ~ access_yes + C(difficulty)",
    "Difficulty + task type adjusted":
        "minutes ~ access_yes + C(difficulty) + C(task_type)"
}

results = []
rq3b_shared_models = {}

for label, formula in specifications.items():
    model = smf.ols(formula, data=rq3b_shared, missing="raise")

    # Check that the model can be estimated with HC3
    assert np.linalg.matrix_rank(model.exog) == model.exog.shape[1]
    leverage = model.fit().get_influence().hat_matrix_diag
    assert leverage.max() < 1 - 1e-8

    fitted = model.fit(cov_type="HC3", use_t=True)
    rq3b_shared_models[label] = fitted
    lower, upper = fitted.conf_int().loc["access_yes"]

    results.append({
        "Model": label,
        "N": int(fitted.nobs),
        "Difference (Yes − No)": fitted.params["access_yes"],
        "95% CI lower": lower,
        "95% CI upper": upper
    })

display(pd.DataFrame(results).round(1))
```

```python
# RQ2b: inspect the original pilot before scoring.
import pandas as pd

if (
    df_filtered.shape[1] != 31
    or "briefly describe" not in str(df_filtered.columns[10]).lower()
    or "suitability" not in str(df_filtered.columns[29]).lower()
):
    raise ValueError("Check column names: the original pilot layout has changed.")

rq2b_raw = df_filtered.iloc[:, 10:30].copy()
rq2b_raw.columns = (
    ["task_description"]
    + [f"skill_{i:02d}" for i in range(1, 17)]
    + ["accuracy", "completeness", "suitability"]
)

# Clean surrounding spaces; keep the original df_filtered unchanged.
rq2b_raw = rq2b_raw.apply(
    lambda s: s.astype("string").str.strip().replace("", pd.NA)
)

print(f"Records being inspected: {len(rq2b_raw)}")

missing = rq2b_raw.isna().sum()
if missing.any():
    display(missing[missing > 0].to_frame("Missing responses"))
else:
    print("No blank responses in the task description, skill, or quality items.")

print("\nObserved response labels across the 16 skill items:")
for response in rq2b_raw.iloc[:, 1:17].stack().unique():
    print(response)

for outcome in ["accuracy", "completeness", "suitability"]:
    print(f"\n{outcome.capitalize()}:")
    display(
        rq2b_raw[outcome]
        .value_counts(dropna=False)
        .rename_axis("Response")
        .to_frame("n")
    )
```

```python
# RQ2b: inspect response distributions for the 16 skill items.
skill_columns = [f"skill_{i:02d}" for i in range(1, 17)]
skill_levels = [
    "Not at all",
    "To a small extent",
    "To a moderate extent",
    "To a large extent",
    "To a very large extent"
]

# Stop if an unexpected label would otherwise be lost during coding.
observed = set(rq2b_raw[skill_columns].stack().dropna().unique())
unexpected = observed - set(skill_levels)
if unexpected:
    raise ValueError(f"Unrecognized skill responses: {sorted(unexpected)}")

rq2b_item_check = (
    rq2b_raw[skill_columns]
    .apply(lambda s: s.value_counts())
    .T.reindex(columns=skill_levels)
    .fillna(0)
    .astype(int)
)

rq2b_item_check.columns = [
    "Not at all", "Small", "Moderate", "Large", "Very large"
]
rq2b_item_check["Missing"] = rq2b_raw[skill_columns].isna().sum()
rq2b_item_check["Total"] = rq2b_item_check.sum(axis=1)

rq2b_item_check.insert(
    0, "Bundle",
    ["Technical/analytical"] * 4
    + ["Evaluation/oversight"] * 3
    + ["Communication"] * 3
    + ["Complex/systems"] * 3
    + ["Learning/self-management"] * 3
)

# Question numbers refer to the updated questionnaire.
rq2b_item_check.index = [f"Q{i}" for i in range(11, 27)]
rq2b_item_check.index.name = "Current question"

display(rq2b_item_check)
```

```python
# RQ2b: calculate the five predefined skill-application scores.
skill_codes = {label: score for score, label in enumerate(skill_levels, 1)}

rq2b_skill_scores = (
    rq2b_raw[skill_columns]
    .apply(lambda s: s.map(skill_codes))
    .astype(float)
)

bundle_items = {
    "Technical/analytical": skill_columns[0:4],
    "Evaluation/oversight": skill_columns[4:7],
    "Communication": skill_columns[7:10],
    "Complex/systems": skill_columns[10:13],
    "Learning/self-management": skill_columns[13:16]
}

# Require every item in a bundle; do not fill missing responses.
rq2b_bundle_scores = pd.DataFrame({
    bundle: rq2b_skill_scores[items].mean(axis=1, skipna=False)
    for bundle, items in bundle_items.items()
})

rq2b_bundle_summary = (
    rq2b_bundle_scores
    .agg(["count", "mean", "std", "min", "max"])
    .T.rename(columns={"count": "n", "std": "SD"})
)
rq2b_bundle_summary["n"] = rq2b_bundle_summary["n"].astype(int)

print("Bundle scores — scale 1 to 5:")
display(rq2b_bundle_summary.round(2))

rq2b_bundle_correlations = rq2b_bundle_scores.corr(method="spearman")

print("\nCorrelations between bundle scores — Spearman:")
display(rq2b_bundle_correlations.round(2))
```

```python
# RQ2b: code the three quality outcomes.
quality_maps = {
    "accuracy": {
        "Not at all accurate": 1,
        "Slightly accurate": 2,
        "Moderately accurate": 3,
        "Mostly accurate": 4,
        "Completely accurate": 5
    },
    "completeness": {
        "0% — None": 0,
        "About 25% — A quarter": 25,
        "About 50% — Half": 50,
        "About 75% — Three quarters": 75,
        "100% — All": 100
    },
    "suitability": {
        "Not at all suitable": 1,
        "Slightly suitable": 2,
        "Moderately suitable": 3,
        "Mostly suitable": 4,
        "Fully suitable": 5
    }
}

rq2b_quality = pd.DataFrame(index=rq2b_raw.index)

for outcome, coding in quality_maps.items():
    unexpected = set(rq2b_raw[outcome].dropna()) - set(coding)
    if unexpected:
        raise ValueError(f"{outcome}: unrecognized responses {unexpected}")

    rq2b_quality[outcome] = rq2b_raw[outcome].map(coding).astype(float)

rq2b_analysis = pd.concat(
    [rq2b_bundle_scores, rq2b_quality], axis=1
)

if rq2b_analysis.isna().any().any():
    raise ValueError("Missing coded values: inspect these before proceeding.")

# Unadjusted associations between each bundle and each quality outcome.
rq2b_quality_correlations = (
    rq2b_analysis.corr(method="spearman")
    .loc[rq2b_bundle_scores.columns, rq2b_quality.columns]
)

print(f"n = {len(rq2b_analysis)} for every association")
print("Bundle–quality associations: Spearman correlations")
display(rq2b_quality_correlations.round(3))
```

```python
# RQ2b: joint associations of all five bundles with each quality outcome.
import numpy as np
from statsmodels.miscmodels.ordinal_model import OrderedModel
from statsmodels.stats.multitest import multipletests

X = rq2b_analysis[rq2b_bundle_scores.columns].astype(float)
outcomes = ["accuracy", "completeness", "suitability"]

if not np.isfinite(rq2b_analysis.to_numpy(dtype=float)).all():
    raise ValueError("Inspect missing or nonfinite values before fitting.")

rq2b_models = {}
results = []

for outcome in outcomes:
    # Use only observed categories, ordered from lower to higher quality.
    y = pd.Categorical(rq2b_analysis[outcome], ordered=True)

    # OrderedModel estimates thresholds; do not add an intercept to X.
    model = OrderedModel(y.codes, X, distr="logit")
    fit = model.fit(method="bfgs", maxiter=1000, disp=False)
    rq2b_models[outcome] = fit

    valid = (
        fit.mle_retvals.get("converged", False)
        and np.isfinite(fit.params).all()
        and np.isfinite(fit.bse).all()
        and (fit.bse > 0).all()
        and np.isfinite(fit.pvalues).all()
    )
    if not valid:
        raise RuntimeError(
            f"{outcome}: inspect model convergence or standard errors."
        )

    ci = fit.conf_int().loc[X.columns]

    for bundle in X.columns:
        results.append({
            "Outcome": outcome,
            "Bundle": bundle,
            "OR": np.exp(fit.params[bundle]),
            "95% CI lower": np.exp(ci.loc[bundle].iloc[0]),
            "95% CI upper": np.exp(ci.loc[bundle].iloc[1]),
            "p_raw": fit.pvalues[bundle]
        })

rq2b_joint_results = pd.DataFrame(results)

estimate_columns = ["OR", "95% CI lower", "95% CI upper"]
if not np.isfinite(rq2b_joint_results[estimate_columns].to_numpy()).all():
    raise RuntimeError("Nonfinite odds ratios or intervals: inspect model fits.")

rq2b_joint_results["Holm p"] = multipletests(
    rq2b_joint_results["p_raw"], method="holm"
)[1]

print(f"All three models converged; n = {len(X)} in each.")
print("Odds ratios represent a one-point increase in a bundle score.")

display(
    rq2b_joint_results.drop(columns="p_raw").style.format({
        "OR": "{:.2f}",
        "95% CI lower": "{:.2f}",
        "95% CI upper": "{:.2f}",
        "Holm p": lambda p: "<0.001" if p < 0.001 else f"{p:.3f}"
    })
)
```

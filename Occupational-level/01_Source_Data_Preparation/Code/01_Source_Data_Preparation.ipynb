The skills data were downloaded from [O*NET Database 30.0 — Skills](https://www.onetcenter.org/dl_files/database/db_30_0_text/Skills.txt)

```python
from pathlib import Path
import pandas as pd
ROOT = Path("/content/drive/MyDrive/GenAI-human-skill-adaptation-dev/Occupational-level/01_Source_Data_Preparation")

skills = pd.read_csv(
    ROOT / "Input" / "Skills.txt",
    sep="\t", keep_default_na=False, na_values=[""]
)
print(f"Rows: {skills.shape[0]} | Columns: {skills.shape[1]}")
```

```python
skills_im = skills.loc[skills["Scale ID"] == "IM"].copy()
skills_im["soc6"] = skills_im["O*NET-SOC Code"].str[:7]

print(
    f"Rows: {len(skills_im)} | "
    f"Detailed occupations: {skills_im['O*NET-SOC Code'].nunique()} | "
    f"Skills: {skills_im['Element ID'].nunique()}"
)
```

The Content Model Reference data were downloaded from [O*NET Database 30.0 — Skills](https://www.onetcenter.org/dl_files/database/db_30_0_text/Content%20Model%20Reference.txt)

```python
content_model = pd.read_csv(
    ROOT / "Input" / "Content Model Reference.txt", sep="\t"
)

skill_reference = content_model.loc[
    content_model["Element ID"].isin(skills_im["Element ID"])
].reset_index(drop=True)

print(f"Skills in reference: {len(skill_reference)}")
```

```python
OUTPUT = ROOT / "Output"

skills_im.to_csv(OUTPUT / "Occupational_Skill_Importance_Long.csv", index=False)
skill_reference.to_csv(OUTPUT / "Skill_Element_Reference.csv", index=False)

print("Saved skill importance ratings and skill definitions.")
```

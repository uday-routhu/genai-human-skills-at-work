**Skill-bundle specification**

This study organizes 26 O*NET skills into five conceptual bundles to examine how occupational human-skill requirements relate to GenAI augmentation share. The selected domains are consistent with research discussing technical, cognitive, interpersonal, and learning capabilities in digital work and AI-related upskilling (van Laar et al., 2017; Jaiswal et al., 2022). The five-bundle structure and individual assignments are study-specific interpretations of O*NET skill definitions.

| Bundle | Included O*NET skills | Grouping rationale |
|---|---|---|
| Technical and analytical skills | Mathematics; Science; Operations Analysis; Technology Design; Programming; Troubleshooting | Combines quantitative and scientific methods with technical design, implementation, and diagnosis. Troubleshooting is included for its emphasis on identifying causes of operating errors. |
| Critical thinking and evaluation | Critical Thinking; Monitoring; Quality Control Analysis; Judgment and Decision Making; Systems Evaluation | Groups skills concerned with assessing alternatives, performance, quality, and corrective action. Quality Control Analysis contributes evaluation of products, services, and processes. |
| Communication and interpersonal skills | Reading Comprehension; Active Listening; Writing; Speaking; Social Perceptiveness; Coordination; Persuasion; Negotiation; Instructing; Service Orientation | Combines receiving and conveying information with understanding, coordinating, influencing, teaching, and assisting others. |
| Complex problem solving and systems analysis | Complex Problem Solving; Systems Analysis | Connects developing solutions to complex problems with understanding how system conditions and relationships affect outcomes. Its emphasis is problem resolution and system understanding. |
| Learning and time management | Active Learning; Learning Strategies; Time Management | Combines using new information and selecting learning approaches with organizing time. It represents learning and work organization, rather than a comprehensive measure of self-management. |

Five bundles reflect these selected functional emphases; the number was not estimated through factor analysis. Each selected skill is assigned once. The cited studies support the relevance of these domains but do not validate this exact grouping or establish that five bundles are optimal.

**Scope of exclusions**

The specification excludes nine skills:

- **Equipment Selection, Installation, Operations Monitoring, Operation and Control, Equipment Maintenance, and Repairing:** these emphasize equipment selection, setup, operation, and upkeep, which fall outside the selected technical-analysis, design, and diagnostic emphasis.
- **Management of Financial Resources, Management of Material Resources, and Management of Personnel Resources:** these concern financial, material, and staffing responsibilities beyond the selected domains.

These exclusions define a focused analysis; they do not imply that the omitted skills are unimportant or unrelated to GenAI. The bundles describe occupational skill requirements, not workers’ AI proficiency or skills acquired through training.

```python
# Assign O*NET skill elements to the five named bundles.
bundles = {
    "Technical and analytical skills": [
        "2.A.1.e", "2.A.1.f", "2.B.3.a", "2.B.3.b", "2.B.3.e", "2.B.3.k"
    ],
    "Critical thinking and evaluation": [
        "2.A.2.a", "2.A.2.d", "2.B.3.m", "2.B.4.e", "2.B.4.h"
    ],
    "Communication and interpersonal skills": [
        "2.A.1.a", "2.A.1.b", "2.A.1.c", "2.A.1.d",
        "2.B.1.a", "2.B.1.b", "2.B.1.c", "2.B.1.d", "2.B.1.e", "2.B.1.f"
    ],
    "Complex problem solving and systems analysis": [
        "2.B.2.i", "2.B.4.g"
    ],
    "Learning and time management": [
        "2.A.2.b", "2.A.2.c", "2.B.5.a"
    ]
}
```

```python
bundle_spec = pd.DataFrame(
    [(bundle, element) for bundle, elements in bundles.items() for element in elements],
    columns=["Bundle", "Element ID"]
).merge(skill_reference, on="Element ID", how="left", validate="one_to_one")
```

```python
# Show the number of skills assigned to each bundle.
bundle_spec.groupby("Bundle", sort=False).size().reset_index(name="Skills")
```

```python
# Save the skill-to-bundle mapping for Phase 03.
bundle_spec.to_csv(
    ROOT / "02_Skill_Bundle_Specification" / "Output" / "Skill_Bundle_Specification.csv",
    index=False
)
```
